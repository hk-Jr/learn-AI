# 🌐 ChromaDB Atlas Studio: Web Browser Explorer

Just like **MongoDB Atlas** gives you a web interface to view collections, inspect documents, and search data visually in your web browser, we now have **ChromaDB Atlas Studio**!

### 🚀 How to Open in Your Browser:
The web server is running locally on your computer at:  
👉 **[http://localhost:5050](http://localhost:5050)**  

(Simply open Google Chrome, Microsoft Edge, or your favorite browser and go to `http://localhost:5050`)

---
### 🎨 What You Can Do in the Browser UI:
1. **Browse Collections**: Switch between collections (`enterprise_profiles`, `online_enterprise_data`) in the left sidebar.
2. **Inspect Every Document**: View all 25 company chunks with color-coded badges for each company:
   - 🔵 **Apex Robotics** (Blue)
   - 🟢 **Solaris Green Energy** (Emerald)
   - 🟣 **MediGen Diagnostics** (Purple)
   - 🔷 **CloudScale Systems** (Cyan)
   - 🟡 **FinEdge Capital** (Amber)
3. **Inspect Vector Coordinates**: Click *"⚡ View 384-D Vector Coordinates"* on any card to reveal the raw floating-point numbers stored in the vector database.
4. **Live Semantic Search**: Type any natural language question into the search bar (e.g. *"Who is facing supply chain bottlenecks in Taiwan?"* or *"Show cancer tests"*) and hit Enter to watch matching documents update instantly with **Cosine Match % Badges**!

## 🛠️ Testing the Web API via Python

The browser UI communicates with ChromaDB via lightweight REST endpoints.  
Let's test the API endpoints right here in Python to see how the browser gets its data:

```python
import urllib.request
import json

# 1. Test fetching collections from the Web Server API
url = "http://localhost:5050/api/collections"
req = urllib.request.urlopen(url)
data = json.loads(req.read().decode('utf-8'))

print("📡 Web Server API Response for /api/collections:")
print(json.dumps(data, indent=2))
```

```python
# 2. Test running a search query through the Web Server API
query = "Taiwan"
search_url = f"http://localhost:5050/api/query?collection=enterprise_profiles&q={query}&n=2"

req = urllib.request.urlopen(search_url)
results = json.loads(req.read().decode('utf-8'))

print(f"📡 Web Server Search API Result for '{query}':")
print(f"Matched ID:       {results['ids'][0][0]}")
print(f"Company:          {results['metadatas'][0][0]['company']}")
print(f"Section:          {results['metadatas'][0][0]['section']}")
print(f"Distance:         {results['distances'][0][0]:.4f}")
```
